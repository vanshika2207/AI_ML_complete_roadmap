```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
data=pd.read_csv('salary_data.csv')
```

```python
data.info()
```

```python
X=data.iloc[:,:-1].values
Y=data.iloc[:,-1].values
```

```python
X,Y
```

```python
X.shape
```

```python
Y.shape
```

```python
plt.scatter(X,Y,color='red')
plt.title('Salary Vs Experience (Observations)')
plt.xlabel('Years of Experience')
plt.ylabel('Salary')
plt.show()
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size=1/3,random_state=0)
```

```python
from sklearn.linear_model import LinearRegression
model=LinearRegression()
model.fit(X_train,Y_train)
```

```python
print('Coefficinet = Theta1 = ', model.coef_)
print('Intercept =Theta0 = ',model.intercept_)
```

```python
plt.scatter(X,Y,color='blue')
plt.plot(X_train,model.predict(X_train),color='green')
plt.title('Salary Vs Experience (Training Set)')
plt.xlabel('Years of Experience')
plt.ylabel('Salary')
plt.show()
```

```python

import seaborn
seaborn.residplot(x=X, y=Y)
plt.show()
```

```python
# Predict salary that can be offered for 7 years of experience
x=np.atleast_1d(7).reshape(-1,1)
```

```python
x
```

```python
# Predict the value 
y_pred = model.predict(x)

print('For x= 7 yrs of experience, predicted salary = ', y_pred)
```

```python
Y_predict = model.predict(X_test)
```

```python
from sklearn.metrics import mean_squared_error,r2_score
# The mean squared error
print("Mean squared error = %.2f" % mean_squared_error(Y_test, Y_predict))

# Explained variance score: 1 is perfect prediction
print('Variance score = %.2f' % r2_score(Y_test, Y_predict))
```
