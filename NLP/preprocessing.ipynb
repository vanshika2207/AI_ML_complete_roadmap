Text Preprocessing
Following code can be used for text preprocessing useful for various NLP applications.

First we need to import nltk

For a given text, we can do sentence tokenization and word tokenization using nltk library functions. We can remove the punctuations using string library.

We can then remove stop words in English to get the important words in the text.

We also perform stemming and lemmatization. Stemming and Lemmatization are two different techniques that help reduce our data space. We don’t need to check every single form of a word for reducing the size of the big data corpus.

```python
import nltk
import string
import re

nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
nltk.download('wordnet')
```

```python
# sentence tokenizer
from nltk.tokenize import sent_tokenize
# word tokenizer
from nltk.tokenize import word_tokenize
```

```python
statement="The QUICK brown foxes were jumping over 5 lazy dogs!!! They aren't running anymore, because the dogs had already been sleeping. #NLP @student"
```

```python
sentence=sent_tokenize(statement)
words=word_tokenize(statement)
```

```python
print(sentence)
```

```python
print(words)
```

```python
only_words=[w for w in words if w not in string.punctuation]
print(only_words)
```

```python
# stop words
from nltk.corpus import stopwords
english_stop_words=set(stopwords.words("english"))
keywords=[ w for w in only_words if w not in english_stop_words]
print(keywords)
```

```python
# Lemmatization
from nltk.stem import WordNetLemmatizer
lemmatizer=WordNetLemmatizer()
lematized_words=[lemmatizer.lemmatize(w) for w in keywords]
print(lematized_words)
```

```python
# Stemming
from nltk.stem import PorterStemmer
porter_stemmer=PorterStemmer()
stemmed_words=[porter_stemmer.stem(w) for w in keywords]
print(stemmed_words)
```
